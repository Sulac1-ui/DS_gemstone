# Introduction About the Data :
The dataset The goal is to predict price of given diamond (Regression Analysis).

There are 10 independent variables (including id):

* id : unique identifier of each diamond
* carat : Carat (ct.) refers to the unique unit of weight measurement used exclusively to weigh gemstones and diamonds.
* cut : Quality of Diamond Cut
* color : Color of Diamond
* clarity : Diamond clarity is a measure of the purity and rarity of the stone, graded by the visibility of these characteristics under 10-power magnification.

* depth : The depth of diamond is its height (in millimeters) measured from the culet (bottom tip) to the table (flat, top surface)

* table : A diamond's table is the facet which can be seen when the stone is viewed face up.
* x : Diamond X dimension
* y : Diamond Y dimension
* x : Diamond Z dimension

```python
df['cut'].value_counts()
df['clarity'].value_counts()
```

```python
cut_map= {"Premium":1, "Very Good": 2,"Good":3, "Fair":4, "Ideal":5 }
Clarity_map= {
    "SI1":1,
    "VS2":2,
    "VS1":3,
    "SI2":4,
    "VVS2":5,
    "VVS1":6,
    "IF":7,
    "I1":8
}
color_mapper = {"D":1 ,"E":2 ,"F":3 , "G":4 ,"H":5 , "I":6, "J":7}
```

```python
df['cut']= df['cut'].replace(cut_map)
df['color']= df['color'].replace(color_mapper)
df['clarity']= df['clarity'].replace(Clarity_map)
```

```python
df= df.drop(columns=['Cut', 'Color', 'Clarity'])
```
